```python
import os
import base64
from google.auth.transport.requests import Request
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from googleapiclient.discovery import build

# Gmail API scope (read-only access)
SCOPES = ['https://www.googleapis.com/auth/gmail.readonly']

SENDER_EMAIL = 'dispatchcounterdept@citizenshospitals.com'
ATTACHMENT_DIR = 'attachments'

def authenticate_gmail():
    creds = None
    if os.path.exists('token.json'):
        creds = Credentials.from_authorized_user_file('token.json', SCOPES)
    if not creds or not creds.valid:
        if creds and creds.expired and creds.refresh_token:
            creds.refresh(Request())
        else:
            flow = InstalledAppFlow.from_client_secrets_file('credentials.json', SCOPES)
            creds = flow.run_local_server(port=0)
        with open('token.json', 'w') as token:
            token.write(creds.to_json())
    return build('gmail', 'v1', credentials=creds)

def get_email_header(payload, header_name):
    headers = payload.get('headers', [])
    for header in headers:
        if header['name'].lower() == header_name.lower():
            return header['value']
    return None

def download_attachments(service, message, msg_id):
    payload = message.get('payload', {})
    parts = payload.get('parts', [])

    os.makedirs(ATTACHMENT_DIR, exist_ok=True)
    saved_files = set(os.listdir(ATTACHMENT_DIR))  # Track existing files

    for part in parts:
        filename = part.get('filename')
        body = part.get('body', {})
        if filename and 'attachmentId' in body:
            # Generate unique filename
            base, ext = os.path.splitext(filename)
            unique_filename = filename
            counter = 1
            while unique_filename in saved_files:
                unique_filename = f"{base}_{counter}{ext}"
                counter += 1
            saved_files.add(unique_filename)

            # Get attachment content
            attachment = service.users().messages().attachments().get(
                userId='me', messageId=msg_id, id=body['attachmentId']
            ).execute()
            file_data = base64.urlsafe_b64decode(attachment['data'].encode('UTF-8'))

            # Save file
            file_path = os.path.join(ATTACHMENT_DIR, unique_filename)
            with open(file_path, 'wb') as f:
                f.write(file_data)
            print(f"✅ Saved attachment: {file_path}")

def fetch_all_emails_with_attachments(service):
    query = f'from:{SENDER_EMAIL} has:attachment'
    next_page_token = None
    total_count = 0

    while True:
        response = service.users().messages().list(
            userId='me', q=query, pageToken=next_page_token
        ).execute()

        messages = response.get('messages', [])
        if not messages:
            break

        for msg in messages:
            msg_id = msg['id']
            msg_data = service.users().messages().get(userId='me', id=msg_id).execute()
            subject = get_email_header(msg_data['payload'], 'Subject') or '(No Subject)'
            print(f"\n📩 Subject: {subject}")
            download_attachments(service, msg_data, msg_id)
            total_count += 1

        next_page_token = response.get('nextPageToken')
        if not next_page_token:
            break

    print(f"\n🎉 Done! Processed {total_count} emails from {SENDER_EMAIL}.\n")

if __name__ == '__main__':
    service = authenticate_gmail()
    fetch_all_emails_with_attachments(service)
```
